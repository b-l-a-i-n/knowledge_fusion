```python
def calc_final_answer(arrs, metric, type):
    msps = []
    for arr in arrs:
        if metric == 'msp':
            m = arr.msp
        elif metric == 'perplexity':
            m = arr.perplexity
        else:
            m = arr.entropy
        msps.append(m)
    msps = np.array(msps)
    if type == 'min':
        position = msps.argmin()
    else:
        position = msps.argmax()
    answers = [arr.answers.strip() for arr in arrs]
    final_answer = answers[position]
    return final_answer
```

```python
def calc_optimal_metric_choose(metric, type):
    acc_sole_max, acc_double_max, acc_mix_max, acc_all_max = 0, 0, 0, 0
    acc_sole_min, acc_double_min, acc_mix_min, acc_all_min = 0, 0, 0, 0
    for idx in range(0, 817):
        arr1 = mistral_w_df.iloc[idx]
        arr2 = mistral_d_df.iloc[idx]
        arr3 = mistral_g_df.iloc[idx]
        arr4 = mistral_wg_df.iloc[idx]
        arr5 = mistral_dw_df.iloc[idx]
        arr6 = mistral_dg_df.iloc[idx]
        arr7 = mistral_dgw_df.iloc[idx]
        
        arrs_sole = [arr1, arr2, arr3]
        arrs_double = [arr4, arr5, arr6]
        arrs_mix = [arr1, arr2, arr3, arr4, arr5, arr6]
        arrs_all = [arr1, arr2, arr3, arr4, arr5, arr6, arr7]
        correct_answer = arr1.correct.strip()
        
        
        sole_final_answer = calc_final_answer(arrs_sole, metric, 'max')
        if sole_final_answer == correct_answer:
            acc_sole_max += 1

        double_final_answer = calc_final_answer(arrs_double, metric, 'max')
        if double_final_answer == correct_answer:
            acc_double_max += 1

        mix_final_answer = calc_final_answer(arrs_mix, metric, 'max')
        if mix_final_answer == correct_answer:
            acc_mix_max += 1

        all_final_answer = calc_final_answer(arrs_all, metric, 'max')
        if all_final_answer == correct_answer:
            acc_all_max += 1

        sole_final_answer = calc_final_answer(arrs_sole, metric, 'min')
        if sole_final_answer == correct_answer:
            acc_sole_min += 1

        double_final_answer = calc_final_answer(arrs_double, metric, 'min')
        if double_final_answer == correct_answer:
            acc_double_min += 1

        mix_final_answer = calc_final_answer(arrs_mix, metric, 'min')
        if mix_final_answer == correct_answer:
            acc_mix_min += 1

        all_final_answer = calc_final_answer(arrs_all, metric, 'min')
        if all_final_answer == correct_answer:
            acc_all_min += 1
    print(f'{acc_sole_max/817} / {acc_sole_min/817}')
    print(f'{acc_double_max/817} / {acc_double_min/817}')
    print(f'{acc_mix_max/817} / {acc_mix_min/817}')
    print(f'{acc_all_max/817} / {acc_all_min/817}')
```

```python
calc_optimal_metric_choose('perplexity', 'max')
```
